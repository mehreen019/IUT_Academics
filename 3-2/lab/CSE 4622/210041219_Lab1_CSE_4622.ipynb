```python
import urllib.request

url ="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

iris = urllib.request.urlopen(url)
print("Primary url info: ", iris)

iris_decoded = iris.read().decode().strip().split('\n')
print("URL decoded: ", iris_decoded)

iris_lines = [line.split(',') for line in iris_decoded]
print("Dataset lines: ", iris_lines)

iris_array = np.array(iris_lines)
print("Dataset as array:\n", iris_array)
```

TASK 5

```python
#mean
iris_mean = np.mean(iris_array[:,0].astype(float))
print("mean: ", iris_mean)

#median
iris_median = np.median(iris_array[:,0].astype(float))
print("median: ", iris_median)

#std
iris_std = np.std(iris_array[:,0].astype(float))
print("standard deviation: ", iris_std)

print("Feature by feature summary:")
for i in range(iris_array.shape[1] - 1):
  col = iris_array[:, i]
  print(f"Col {i+1}: mean={np.mean(iris_array[:,i].astype(float))}, median={np.median(iris_array[:,i].astype(float))}, standard deviation={np.std(iris_array[:,i].astype(float))}")
```

**PANDAS**

TASK 1

```python
df = pd.read_csv("sample_data/california_housing_test.csv")
print("First 10 rows: \n", df.head(10))
print("\n Data types: \n", df.dtypes)
print("\n Summary for numerical columns: \n", df.describe())
```

TASK 2

```python
missing = df.isnull().sum()
print("Missing values: \n", missing)

new_df = df.dropna()
print("\n After dropping null rows: \n", new_df)

df_cleaned = df.fillna(df.mean(numeric_only=True))
print("\n Filling missing values with mean: \n", df_cleaned)
```

TASK 3

```python
df['region'] = df['longitude'].round(1).astype(str) + ", " + df['latitude'].round(1).astype(str)
print("After adding region column:")
print(df.head())

region_stats = df.groupby('region').agg({
    'median_house_value': 'mean',
    'median_income': 'mean',
    'housing_median_age': 'mean'
})

region_stats.columns = ['Avg House Value', 'Avg Income', 'Avg House Age']

region_sorted = region_stats.sort_values(
    by=['Avg House Value', 'Avg Income', 'Avg House Age'],
    ascending=[False, False, True]
)

print("\nTop 5 Investment Regions (sorted by highest value, income, and youngest houses):")
print(region_sorted.head())
```
