##Task 1
Complete the following function named inverse_conv():

```python
import numpy as np
import matplotlib.pyplot as plt

def build_conv_matrix(h, x_len):
  N = len(h)
  M = N + x_len - 1
  H = np.zeros((M, x_len))

  for i in range(M):
    for j in range(x_len):
      if 0 <= i - j < N:
        H[i, j] = h[i-j]
  return H

def inverse_conv(y,h):
  h_len = len(h)
  x_len = len(y) - len(h) + 1

  H = build_conv_matrix(h, x_len)
  x = np.dot(np.linalg.pinv(H), y)

  return x

y = np.array([-8, 10, -15, 4, -8, -9, -2, -8])
h = np.array([-2,1,-2])

print(inverse_conv(y,h))
```

##Task 2
Write a function named forwardDFT() that takes a time domain signal $x$ and the number of points N and returns the frequency domain signal as two separate arrays $ReX$ and $ImX$


```python
def forwardDFT(x,N):
  #your code here
  ReX = np.zeros(int(N/2) + 1)
  ImX = np.zeros(int(N/2) + 1)

  for k in range(int(N/2) +1):
    for n in range(len(x)):
      angle = -2 * np.pi * k * n / N
      ReX[k] += x[n] * np.cos(angle)
      ImX[k] += x[n] * np.sin(angle)

  return ReX, ImX


x = np.array([11,6,20,2,40,10,43,20])
N = 8
print(forwardDFT(x,N))
```

Write a function named inverseDFT() that takes a frequency domain signal as two separate arrays $ReX$ and $ImX$ returns the time domain signal $x$.

```python
def inverseDFT(ReX,ImX):
  #your code here
  N_prime = (len(ReX) - 1) * 2

  for k in range(len(ReX)):
    ReX[k] = ReX[k] / (N_prime / 2)
    ImX[k] = -ImX[k] / (N_prime / 2)

  ReX[0] = ReX[0] / 2
  ReX[-1] = ReX[- 1] / 2

  xx = np.zeros(N_prime)

  for i in range(N_prime):
    for k in range(len(ReX)):
      angle = 2 * np.pi * k * i / N_prime

      xx[i] += ReX[k] * np.cos(angle)
      xx[i] += ImX[k] * np.sin(angle)

  return xx

ReX = np.array([152, -19.101, -12, -38.89, 76])
ImX = np.array([0, 38.556, 6, -7.44, 0])
print(inverseDFT(ReX,ImX))
```
